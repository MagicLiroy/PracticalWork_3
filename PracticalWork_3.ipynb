Задание №5

```python
import csv
from collections import defaultdict, Counter

def extract_first_name(name, sex):
    if sex == 'female' and '(' in name:
        start = name.find('(') + 1
        end = name.find(')', start)
        if end != -1:
            name_in_brackets = name[start:end]
            first_name = name_in_brackets.strip().split()[0]
            first_name = first_name.replace('"', '')
            return first_name
    parts = name.split(',', 1)
    rest = parts[1] if len(parts) > 1 else name
    title_parts = rest.split('.', 1)
    after_title = title_parts[1] if len(title_parts) > 1 else rest
    words = after_title.split()
    if words:
        first_name = words[0].replace('"', '')
        return first_name
    return rest.strip().replace('"', '')

counters = defaultdict(Counter)
with open('train.csv', 'r', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for row in reader:
        pclass = row['Pclass']
        sex = row['Sex']
        name = row['Name']
        first_name = extract_first_name(name, sex)
        counters[(pclass, sex)][first_name] += 1

for pclass in ['1', '2', '3']:
    print(f"Class {pclass}:")
    for sex in ['male', 'female']:
        key = (pclass, sex)
        if key in counters:
            most_common = counters[key].most_common(1)
            if most_common:
                name, count = most_common[0]
                print(f"  {sex.capitalize()}: {name} (count: {count})")
```
